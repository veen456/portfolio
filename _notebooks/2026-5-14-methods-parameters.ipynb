## Methods and Parameters

### What are Methods?

A **method** is a reusable block of code that performs a specific task. Instead of writing the same code multiple times, you can create a method once and call it whenever you need it.

**Key points:**
- Methods make code organized and easier to read
- Methods reduce repetition (DRY - Don't Repeat Yourself)
- Methods can be reused throughout your program

### What are Parameters?

**Parameters** are inputs that you pass to a method. They allow you to customize what a method does without rewriting it.

**Key points:**
- Parameters go inside the parentheses when you define a method
- You can pass values (called **arguments**) when you call the method
- Parameters let the same method work with different data

### Simple Example

```python
# Defining a method with parameters
def greet(name):
    print(f"Hello, {name}!")

# Calling the method with different arguments
greet("Alice")    # Output: Hello, Alice!
greet("Bob")      # Output: Hello, Bob!
```

In this example:
- `greet` is the **method** name
- `name` is the **parameter**
- `"Alice"` and `"Bob"` are the **arguments** (actual values passed in)

### Benefits

- **Reusability**: Write once, use many times
- **Flexibility**: Use parameters to change behavior
- **Maintainability**: Easier to fix bugs or make changes
- **Clarity**: Code is more readable and organized
